```python
# CONSTANTES
TemperaturaMinimaValida = -10.0
TemperaturaMaximaValida = 60.0
HumedadMinimaValida = 0.0
HumedadMaximaValida = 100.0

TemperaturaMinimaOptima = 18.0
TemperaturaMaximaOptima = 26.0
HumedadMinimaOptima = 60.0
HumedadMaximaOptima = 80.0

TemperaturaCriticaMinima = 12.0
TemperaturaCriticaMaxima = 32.0
HumedadCriticaMinima = 40.0
HumedadCriticaMaxima = 90.0

MaximoLecturas = 8

# VARIABLES
LecturasContadas = 0
EstadoFinal = ""

ListaTemperaturas = []
ListaHumedades = []

SumaTemperatura = 0.0
SumaHumedad = 0.0

# CICLO PRINCIPAL
while LecturasContadas < MaximoLecturas:
    print(f"\n--- Lectura {LecturasContadas + 1} ---")

    # 1. Leer datos
    Temperatura = float(input("Ingrese la temperatura (°C): "))
    Humedad = float(input("Ingrese la humedad (%): "))

    # 2. Validar que los datos pertenezcan al rango válido del sensor
    if Temperatura < TemperaturaMinimaValida or Temperatura > TemperaturaMaximaValida or Humedad < HumedadMinimaValida or Humedad > HumedadMaximaValida:
        print("Dato inválido. Ingrese nuevamente la lectura (no se cuenta).")
        continue  # Vuelve a pedir la lectura sin contar esta

    # Si es válido, se registra
    LecturasContadas = LecturasContadas + 1
    ListaTemperaturas.append(Temperatura)
    ListaHumedades.append(Humedad)
    SumaTemperatura = SumaTemperatura + Temperatura
    SumaHumedad = SumaHumedad + Humedad

    # 3. Evaluar condiciones
    # ¿Es crítico?
    EsCritico = (Temperatura < TemperaturaCriticaMinima or Temperatura > TemperaturaCriticaMaxima or Humedad < HumedadCriticaMinima or Humedad > HumedadCriticaMaxima)

    # ¿Es óptimo?
    TemperaturaOptima = (Temperatura >= TemperaturaMinimaOptima and Temperatura <= TemperaturaMaximaOptima)
    HumedadOptima = (Humedad >= HumedadMinimaOptima and Humedad <= HumedadMaximaOptima)
    EsOptimo = TemperaturaOptima and HumedadOptima

    if EsCritico:
        EstadoFinal = "Alerta crítica"
        print("Estado: Crítico")
        print("Acción: Alerta crítica: activar protocolo de emergencia")
        break

    elif EsOptimo:
        print("Estado: Óptimo")
        if LecturasContadas == 1:
            EstadoFinal = "Condiciones óptimas"
            print("Acción: Cultivo en condiciones óptimas")
        else:
            EstadoFinal = "Cultivo estabilizado"
            print("Acción: Cultivo estabilizado")
        break

    else:
        # Está en observación
        print("Estado: Observación")
        Acciones = []

        # Determinar acciones correctivas necesarias
        if Temperatura > TemperaturaMaximaOptima:
            Acciones.append("Ventilar")
        elif Temperatura < TemperaturaMinimaOptima:
            Acciones.append("Encender calefacción")

        if Humedad > HumedadMaximaOptima:
            Acciones.append("Deshumidificar")
        elif Humedad < HumedadMinimaOptima:
            Acciones.append("Nebulizar")

        print("Acción: " + " y ".join(Acciones))

        # Si llegó a la octava lectura sin estabilizarse ni ser crítico
        if LecturasContadas == MaximoLecturas:
            EstadoFinal = "Revisión técnica"

# MOSTRAR RESULTADOS FINALES
print("\n" + "="*40)
print("RESUMEN FINAL")
print("="*40)

# Calcular Mínimos y Máximos
TemperaturaMinima = min(ListaTemperaturas)
TemperaturaMaxima = max(ListaTemperaturas)
PromedioTemperatura = SumaTemperatura / LecturasContadas

HumedadMinima = min(ListaHumedades)
HumedadMaxima = max(ListaHumedades)
PromedioHumedad = SumaHumedad / LecturasContadas

print(f"Resultado: {EstadoFinal}. Lecturas tomadas: {LecturasContadas}.")
print(f"Temperatura: mínima {TemperaturaMinima} – máxima {TemperaturaMaxima} – promedio {PromedioTemperatura:.2f}.")
print(f"Humedad: mínima {HumedadMinima} – máxima {HumedadMaxima} – promedio {PromedioHumedad:.2f}.")

# Formatear secuencias separadas por punto y coma
SecuenciaTemperaturas = "; ".join(str(x) for x in ListaTemperaturas)
SecuenciaHumedades = "; ".join(str(x) for x in ListaHumedades)

print(f"Secuencia de temperaturas: {SecuenciaTemperaturas}.")
print(f"Secuencia de humedades: {SecuenciaHumedades}.")
```
